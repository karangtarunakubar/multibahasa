## Language composition
Check that repository language percentages add up to 100%.


```python
language_rows = repo_data['language_composition']
assert abs(sum(item['percent'] for item in language_rows) - 100) < 1e-9
[(item['name'], item['percent']) for item in language_rows]
```

## Create a local SQLite dictionary
The page `language-data.html` generates SQL inserts, while this cell shows the target table and a safe parameterized insert pattern. The browser page does not connect to or execute SQL against a server.


```python
db = sqlite3.connect(':memory:')
db.execute('''CREATE TABLE language_words (
    id INTEGER PRIMARY KEY,
    language_code TEXT NOT NULL,
    language_name TEXT NOT NULL,
    word TEXT NOT NULL,
    translation TEXT NOT NULL,
    created_at TEXT NOT NULL DEFAULT CURRENT_TIMESTAMP,
    UNIQUE(language_code, word)
)''')
sample = ('tj-TJ', 'Tunjung', 'Emooq', 'tinggi')
db.execute(
    'INSERT INTO language_words (language_code, language_name, word, translation) VALUES (?, ?, ?, ?)',
    sample,
)
db.execute('SELECT language_name, word, translation FROM language_words').fetchall()
```
